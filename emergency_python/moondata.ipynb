```python
import pandas as pd
import geopandas as gpd
import requests
from urllib.parse import urlencode
```

```python
df = pd.read_csv('moonlight.csv',encoding='cp949')
```

```python
df
```

```python
# It is possible to create a geometry column with the given coordinates
# gpd.points_from_xy is a function to create points based on given coordinates

moonlight_gdf = gpd.GeoDataFrame(df, 
                                # a function to create points based on given coordinates
                                geometry=gpd.points_from_xy(df.longitude, df.latitude) 
                               )
moonlight_gdf
```

```python
print(moonlight_gdf.crs)
```

```python
# .set_crs() method is used to set the CRS of the GeoDataFrame
moonlight_gdf = moonlight_gdf.set_crs(epsg=4326)
```

```python
moonlight_gdf.crs
```

```python
# With the correct crs, the .explore() method works
moonlight_gdf.explore()
```

```python
moonlight_gdf.to_file("moonlight_gdf.geojson", driver="GeoJSON", encoding="utf-8")
```
