```python
import pandas as pd
import requests
from urllib.parse import urlencode
```

```python
# 소아과 병원 리스트
hospital_child = []
```

```python
# 소아과 병원 디테일 정보 저장용
hospital_child_detail = []
```

```python
# pediatrics hospital
request_url = f'https://icaretok.nemc.or.kr/pec/portal/med/search/medSearchPaging.do'
```

```python
# pediatrics hospital detail
request_url_detail = f'https://icaretok.nemc.or.kr/pec/portal/med/search/medSearchDetail.do'
```

```python
for page in range(1,17):
    # moonlight hospital params (for POST request)
    params = {
    'page': page,
    'rows': '1000',
    'medGbn': 'hospital',
    'tabGbn': 'hospital',
    'hospitalTab': 'pediatrics'}
    # send request
    response = requests.post(url= request_url, data= params)
    # response data
    data = response.json()
    hospital_children = data['data']
    # data 길이만큼 반복
    for i in range(0,len(data['data'])):
        hospital_child.append(data['data'][i])
    
```

```python
for page in range(1,17):
    # moonlight hospital params (for POST request)
    params = {
    'page': page,
    'rows': '1000',
    'medGbn': 'hospital',
    'tabGbn': 'hospital',
    'hospitalTab': 'pediatrics'}
    # send request
    response = requests.post(url= request_url, data= params)
    # response data
    data = response.json()
    hospital_children = data['data']
    # data 길이만큼 반복
    for i in range(0,len(data['data'])):
        # moonlight hospital detail params (for POST request)
        params_detail = {
            'hpId': data['data'][i]['hpid'],
            'medGbn': 'hospital'}
        # send request
        response_detail = requests.post(url= request_url_detail, data= params_detail)
        data_detail = response_detail.json()['medInfo']
        hospital_child_detail.append(data_detail)
        hospital_child.append(data['data'][i])


    
    
```

```python
df_hospital_children = pd.DataFrame(hospital_child)
```

```python
df_hospital_children_datail = pd.DataFrame(hospital_child_detail)
```
