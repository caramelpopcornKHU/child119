```python
import geopandas as gpd
from sqlalchemy import create_engine
from urllib.parse import quote_plus
```

```python
# 1. GeoJSON 파일 불러오기
gdf = gpd.read_file("emergency.geojson")
```

```python
# 2. geometry를 WKT로 변환
gdf["geometry_wkt"] = gdf["geometry"].apply(lambda x: x.wkt)
```

```python
# 4. geometry 컬럼 제외 후 MySQL에 저장
gdf.drop(columns=["geometry"], inplace=True)
gdf.to_sql("emergency", con=engine, if_exists="replace", index=False)

print("✅ Data inserted into MySQL table 'emergency'")
```
