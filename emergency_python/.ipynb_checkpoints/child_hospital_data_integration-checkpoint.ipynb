```python
import pandas as pd
import geopand as gpd
```

```python
df_c = pd.read_csv('data_hospital_geo_clear/child_c.csv',encoding='utf-8')
```

```python
df_gh = pd.read_csv('data_hospital_geo_clear/child_gh.csv',encoding='utf-8')
```

```python
df_h = pd.read_csv('data_hospital_geo_clear/child_h.csv',encoding='utf-8')
```

```python
df_tgh = pd.read_csv('data_hospital_geo_clear/child_tgh.csv',encoding='utf-8')
```

```python
# 4개의 DataFrame을 리스트에 담기
all_dfs = [df_c, df_gh, df_h, df_tgh]

# pd.concat을 사용하여 모든 DataFrame을 한 번에 합치기
combined_df = pd.concat(all_dfs)

print("--- 인덱스 유지 (기본) ---")
print(combined_df)
```
