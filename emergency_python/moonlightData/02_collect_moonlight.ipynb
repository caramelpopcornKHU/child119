```python
import pandas as pd
import requests
from urllib.parse import urlencode
```

```python
# 달빛어린이 병원 리스트
hospital_moonlight = []
```

```python
# 달빛어린이 병원 디테일 정보 저장용
hospital_moonlight_detail = []
```

```python
# moonlight hospital
request_url = f'https://icaretok.nemc.or.kr/pec/portal/med/search/medSearchPaging.do'
```

```python
# moonlight hospital detail
request_url_detail = f'https://icaretok.nemc.or.kr/pec/portal/med/search/medSearchDetail.do'
```

```python
# moonlight hospital params (for POST request)
params = {
    'page': '1',
    'rows': '130',
    'medGbn': 'hospital',
    'tabGbn': 'hospital',
    'hospitalTab': 'moonLight'
}
```

```python
# send request
response = requests.post(url= request_url, data= params)
```

```python
# response data
data = response.json()
```

```python
hospital_moonlight = data['data']
```

```python
# data 길이만큼 반복
for i in range(0,len(data['data'])):

    # moonlight hospital detail params (for POST request)
    params_detail = {
    'hpId': data['data'][i]['hpid'],
    'medGbn': 'hospital'}
    # send request
    response_detail = requests.post(url= request_url_detail, data= params_detail)
    data_detail = response_detail.json()['medInfo']
    hospital_moonlight_detail.append(data_detail)

```

```python
df_hospital = pd.DataFrame(hospital_moonlight)
```

```python
df_datail = pd.DataFrame(hospital_moonlight_detail)
```
