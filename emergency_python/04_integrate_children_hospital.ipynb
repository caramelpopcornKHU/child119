```python
import pandas as pd
import geopandas as gpd
```

```python
df_c = pd.read_csv('data_hospital_geo_clear/child_c.csv',encoding='utf-8')
```

```python
df_gh = pd.read_csv('data_hospital_geo_clear/child_gh.csv',encoding='utf-8')
```

```python
df_h = pd.read_csv('data_hospital_geo_clear/child_h.csv',encoding='utf-8')
```

```python
df_tgh = pd.read_csv('data_hospital_geo_clear/child_tgh.csv',encoding='utf-8')
```

```python
# 4개의 DataFrame을 리스트에 담기
all_dfs = [df_c, df_gh, df_h, df_tgh]

# pd.concat을 사용하여 모든 DataFrame을 한 번에 합치기
combined_df = pd.concat(all_dfs)

print("--- 인덱스 유지 (기본) ---")
print(combined_df)
```

```python
combined_df
```

```python
total_df = combined_df.iloc[:,1:]
```

```python
total_df
```

```python
total_df.to_csv('child_hs')
```

```python
# It is possible to create a geometry column with the given coordinates
# gpd.points_from_xy is a function to create points based on given coordinates

total_hospital_gdf = gpd.GeoDataFrame(total_df, 
                                # a function to create points based on given coordinates
                                geometry=gpd.points_from_xy(total_df.x, total_df.y) 
                               )
```

```python
total_hospital_gdf
```

```python
print(total_hospital_gdf.crs)
```

```python
# .set_crs() method is used to set the CRS of the GeoDataFrame
total_hospital_gdf = total_hospital_gdf.set_crs(epsg=4326)
```

```python
print(total_hospital_gdf.crs)
```

```python
total_hospital_gdf.crs
```

```python
# With the correct crs, the .explore() method works
total_hospital_gdf.explore()
```

```python
total_hospital_gdf.columns
```

```python
# 변경할 이름 (DB 기준: snake_case)
new_columns = {
    '병원/약국명': 'name',
    '병원/약국구분': 'type',
    '전화번호': 'phone_number',
    '우편번호': 'postal_code',
    '소재지주소': 'address',
    '홈페이지': 'website',
    'x': 'longitude',
    'y': 'latitude',
    'geometry': 'geometry' # 이름이 같아도 명시적으로 포함 가능
}
# 2. 컬럼 이름 변경
total_hospital_gdf_renamed = total_hospital_gdf.rename(columns=new_columns)
```
