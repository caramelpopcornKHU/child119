```python
import requests
from urllib.parse import urlencode
import pandas as pd
import geopandas as gpd
```

```python
import os
# Define necessary variables
YOUR_API_KEY = os.environ.get("VWORLD_API_KEY") # Replace with your actual API key
```

```python
df_c = pd.read_excel('data_hospital/child_c.xlsx')
```

```python
df_c
```

```python
# The destination URL for the geocoding API
base_url = 'https://api.vworld.kr/req/address?'
query_type = 'ROAD' # 'ROAD' (도로명주소) or 'PARCEL' (지번주소)
```

```python
for idx,row in df_c.iterrows():
    # 각 행의 주소값
    query_addr = row['소재지주소']
    # The parameters for the API request
    request_url = f'{base_url}service=address&request=GetCoord&key={YOUR_API_KEY}&type={query_type}&address={query_addr}'
    response = requests.get(request_url)
    #print(response)
    data = response.json()
    if data['response']['status'] == 'OK':
        df_c.loc[idx,'x'] = data['response']['result']['point']['x']
        df_c.loc[idx,'y'] = data['response']['result']['point']['y']
```

```python
df_c
```

```python
df_null = df_c[df_c['x'].isnull()]
```
